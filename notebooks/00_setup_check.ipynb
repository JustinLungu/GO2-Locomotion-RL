## 0.1 GPU check

Training runs thousands of robots in parallel on the GPU.
This cell checks whether PyTorch can see the GPU.
If this returns False, the GPU will **not** be used, and training will be (unusably) slow.
This is fine for analysis, but for training make sure the GPU is available.

````python
import torch

print("CUDA available:", torch.cuda.is_available())
````

## 0.2 Simulation and viewer check

This cell builds a minimal mjlab environment:
16 Unitree Go2 robots on flat ground, with no rewards, observations or terminations yet (you will add those in your notebooks).
The `"zero"` policy sends all-zero actions, so the robots do nothing useful: we only check the simulation runs and can be displayed.

The viewer opens through the printed link.
You can open the page and rotate/zoom the camera with the mouse.

**The cell keeps running while the viewer is open.**
When you are done, stop it with the **interrupt** button (■) in the toolbar.
You **must** do this before running any other cell.

The first run takes a minute or two to initialize because the simulation is compiled for GPU.
Subsequent runs will be faster thanks to caching.

````python
import lr_project # Before mjlab import
from mjlab.envs import ManagerBasedRlEnvCfg
from lr_project import scene, play

# A minimal environment config spawning 16 parallel robots
env_cfg = ManagerBasedRlEnvCfg(
    scene=scene.make_scene_cfg(num_envs=16, terrain="flat"),
    sim=scene.make_sim_cfg(timestep_s=0.005),
    decimation=4,
)

# Interactively visualize the environment. Stop with the interrupt button (■).
play.interactive(env_cfg=env_cfg, policy="zero")
````
